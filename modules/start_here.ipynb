# Welcome to PS4001/PS7001
## Applied Genomics Using Next Generational Sequencing Techniques


This is a project-based course to help students develop skills for analyzing high-
throughput sequencing data, with a focus on RNA-seq, ChIP-seq, and DAP-seq. Topics
include use of the command line, configuration of HPC environments for use with Mizzou’s
Hellbender, and the application of statistical and computational methods utilizing R at a
fundamental level for use in processing NGS data sets. Students will engage in hands-on
projects using sequencing data and current bioinformatics tools to build skills in data
preprocessing, analysis, visualization, and managing workflows.

**NOTE: Please complete [Module_0](module_0_orientation/README.md) and the RSS Intro to Supercomputing (an MU Canvas course; see the [Hellbender documentation](https://docs.itrss.umsystem.edu/pub/hpc/hellbender)) mini-course in canvas before starting on the weekly Modules.**

# Course Module Schedule
    
### [Module 0: Start Here](module_0_orientation/README.md)
Get oriented with course objectives, computing resources, and using Jupyter via RSS OnDemand while completing the RSS Intro to Supercomputing course.

### [Module 1 - Introduction to the Hellbender HPC](module_1_bash_hpc/README.md) 
Gain hands-on experience navigating Linux, writing basic Bash scripts, managing environments, and using command-line tools within Jupyter and the HPC environment.

### [Module 2: File Formats and NGS Techniques](module_2_ngs_preprocessing/README.md)
Explore sequencing data structures, FASTQ preprocessing, trimming, and QC while learning key Bash and AWK skills for handling raw NGS data.

### [Module 3: RNA-Seq Using Pseudo-Alignment](module_3_rnaseq_pseudoalignment/README.md)
Learn how to source, preprocess, and analyze RNA-Seq data using Salmon and DESeq2, focusing on pseudo-alignment and biological interpretation.

### [Module 4: Visualizing Results in R](module_4_rnaseq_in_r/README.md)
Develop R skills for differential expression analysis, graphing, and interpreting biological data through PCA, UpSet, and DESeq2 visualization tools.

### [Module 5: Mapping Reads to a Genome and Using Genome Browsers](module_5_genome_alignment/README.md)
Understand genome alignment workflows, BAM file preparation, and how to visualize mapped reads using IGV for biological interpretation.

### [Module 6: RNA-Seq – Alignment to the Genome](module_6_full_rnaseq_workflow/README.md)
Perform end-to-end RNA-Seq workflows including alignment with STAR, QC, normalization, and visualization using differential expression metrics.

### [Module 7: ChIP-seq and DAP-seq](module_7_dap_seq/README.md)
Apply NGS alignment and peak-calling methods to protein-DNA interaction data, using MACS2, BEDtools, MEME Suite, and R tools for motif and enrichment analysis.

### [Module 8: Final Project Development](module_8_capstone/README.md)
Complete a full RNA-Seq, ChIP-seq, or DAP-seq pipeline analysis using your chosen dataset, interpret results, and present findings in a recorded talk and summary report.
