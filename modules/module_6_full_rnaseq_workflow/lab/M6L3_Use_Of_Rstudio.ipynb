```
    #!/bin/bash
    #load the R-Studio Module
    ml rstudio

    #Path to this environment's R binary
    export RSTUDIO_WHICH_R="$CONDA_PREFIX/bin/R"
    
    #R library path inside the environment
    export R_LIBS_SITE="$CONDA_PREFIX/lib/R/library"
    
    #Ensure R uses conda's version of R's compilers
    export LD_LIBRARY_PATH="$CONDA_PREFIX/lib:$LD_LIBRARY_PATH"
    
    echo "Activated conda env: $(basename $CONDA_PREFIX)"
    echo "RStudio will use: $RSTUDIO_WHICH_R"
```




## R-Studio Interface overview
RStudio is an Integrated Development Environment (IDE) for the R programming language. It includes:
- A text editor for writing scripts
- A console for running R commands
- An environment pane to view variables
- A plots pane for visualizations
- Package management tools
- Project-based workflows

### **2.1 Console**
Where R commands execute immediately.
*[Screenshot omitted from the public version (it showed account-specific details).]*

### **2.2 Source Editor**
Where longer scripts (`.R` files) are written. Open a new `.R` file with the `+` button in the top right hand corner:

![image.png](images/M6L3_Use_Of_Rstudio_0f064d49.png)

*[Screenshot omitted from the public version (it showed account-specific details).]*

### **2.3 Environment/History Pane**
- Shows variables, functions, and datasets currently loaded.
- Helps visualize what is stored in memory.

![image.png](images/M6L3_Use_Of_Rstudio_9f9c6510.png)

### **2.4 Files/Plots/Packages/Help Pane**

*[Screenshot omitted from the public version (it showed account-specific details).]*

- **Files**: Navigate directories
- **Plots**: View `ggplot2` or base R plots
- **Packages**: Install, update, and load packages
- **Help**: Built‑in documentation viewer

## You can navigate each section easily with the mouse, and if there are any questions on how to use each section, visit the documentation here: [R-Studio Documentation](https://docs.posit.co/ide/user/)

Or check out this video: 

<iframe width="560" height="315" src="https://www.youtube.com/embed/aCUM4r1ONhg?si=lv7mhZ6SW_Sd4soe" title="YouTube video player" frameborder="0" allow="accelerometer; autoplay; clipboard-write; encrypted-media; gyroscope; picture-in-picture; web-share" referrerpolicy="strict-origin-when-cross-origin" allowfullscreen></iframe>

## Running Code in R Studio: 
in RStudio, you can run R code in two ways:

### **Console (immediate execution)**
Type commands directly and press Enter.

*[Screenshot omitted from the public version (it showed account-specific details).]*

### **Scripts (for real work)**
- Open a script with: `File > New File > R Script` or `Ctrl + Shift + N`
- Write code
- Run a line with **Ctrl+Enter** to output to the console, or to a file, etc...
- Save the file as `FILENAME.R`

*[Screenshot omitted from the public version (it showed account-specific details).]*


## Installing and loading Packages

In RStudio you install packages using either the **Packages pane** or via code:

*[Screenshot omitted from the public version (it showed account-specific details).]*

or 


```r
install.packages("tidyverse")
library(tidyverse)
```

## Projects in R Studio

RStudio Projects keep your work organized.
- Create with: `File → New Project`
- Keeps working directory consistent
- Bundles scripts, data, and outputs

We can use these packages to keep a project intact, and if you would like, you can save all the RStudio projects within the project folder for your work, or within a central location on your data storage. 

### We can also save R sessions and load them later to avoid having to re-run code to call objects in your environment (plots, data frames, models, variables, etc.)

To save a session: 
```r
save.image("MySession.RData")
```

To load an R session: 
```r
load("MySession.RData")
```


We can also save individual objects with `saveRDS(objectName, "ObjectName.rds")` or a collection of objects with `save(Object1, Object2, Object3, file = "path/to/save/file/at/OBJECTS.RData")`

We can load these with the `load()` function

| Goal                            | Best method     | Commands                              |
| ------------------------------- | --------------- | ------------------------------------- |
| Save everything at once         | `.RData`        | `save.image("file.RData")`            |
| Save specific objects           | `.RData`        | `save(obj1, obj2, file="file.RData")` |
| Save one object cleanly         | `.rds`          | `saveRDS(obj,"obj.rds")`              |
| Reload one object               | `.rds`          | `obj <- readRDS("obj.rds")`           |
| Long-term reproducible analysis | RStudio Project | Use `.Rproj`                          |

- We need to create the metadata in CSV format for this sample set, so visit the [GSE79275](https://www.ncbi.nlm.nih.gov/geo/query/acc.cgi?acc=GSE79275), download the metadata, and parse the file for sample conditions and names. Create a CSV file that looks like to look like this:

5. Load packages in one script called `01_load_packages.R` in `/scripts`
    ```r
    packages <- c("tidyverse", "DESeq2", "pheatmap", "ggplot2")
    
    lapply(packages, require, character.only = TRUE)
    
    ```
6. Import Count data from `/data/counts/CountFile.tsv` with a script in `scripts/02_import_counts.R`.
    ```r
    metadata <- read.csv("data/metadata/samples.csv")
    
    counts <- read.delim("data/counts/gene_counts.tsv", row.names = 1)
    
    # Ensure columns match metadata sample order
    counts <- counts[, metadata$sample]
    
    all(colnames(counts) == metadata$sample) # should be TRUE
    ```
7. Run DESeq2 in `scripts/03_deseq2_analysis.R`
    ```r
    dds <- DESeqDataSetFromMatrix(
        countData = counts,
        colData = metadata,
        design = ~ condition
    )
    
    dds <- DESeq(dds)
    
    res <- results(dds, contrast = c("condition", "KO", "WT"))
    
    write.csv(as.data.frame(res), "results/DEGs/deseq2_results.csv")

    ```

8. QC and Plotting in `scripts/04_visualizations.R`
    ```r
    vsd <- vst(dds)
    
    # PCA Plot
    pca <- plotPCA(vsd, intgroup = "condition")
    ggsave("results/QC/pca_plot.png", pca)
    
    # Heatmap of top variable genes
    topVar <- head(order(rowVars(assay(vsd)), decreasing = TRUE), 100)
    pheatmap(assay(vsd)[topVar, ],
             annotation_col = metadata,
             filename = "results/QC/heatmap.png")
    ```
