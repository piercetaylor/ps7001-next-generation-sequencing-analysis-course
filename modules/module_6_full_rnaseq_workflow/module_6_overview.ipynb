# Module 6 Checklist
## Module 6 - Alignment to the Genome
This week, you will focus on reviewing computational techniques and applying the STAR pipeline to a new dataset, [GSE72548 dataset](https://www.ncbi.nlm.nih.gov/geo/query/acc.cgi?acc=GSE72548). The paper, [The Role of Cytokinin During Infection of Arabidopsis thaliana by the Cyst Nematode *Heterodera schachtii*](https://apsjournals.apsnet.org/doi/pdf/10.1094/MPMI-07-15-0156-R), is avaiable to you on canvas as well. You will spend the majority of your time completeing the exercise, and will run the full STAR pipline on your own. 

**NOTE: This module includes running programs like STAR and Fastp on large file sets (>10GB), so allot time for Jupyter Notebook and Hellbender Desktop sessions in larger increments (4-8 hours) and expect downtime while programs run as you work through the labs, practices, and exercises.**

**Please make sure you have created each Session with CPU cores = 8 and RAM (memory) >= 32 GB for use of the exercise**

## Module Topics:

- Review of computation resources and methods
- Project management and scripting overview
- Use of RStudio
- DESeq2 study design considerations
- Use of STAR for genome alignment
