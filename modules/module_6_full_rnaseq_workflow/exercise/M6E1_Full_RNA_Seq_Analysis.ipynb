
# STAR Pipeline: 

It is expected that you follow the STAR pipeline outlined in Modules 4 and 5 within this course: 
- Create a project folder `m6_project` within `Module_6/` that looks something like this:
    ```
    m6_project/
    ├── data
    │   ├── raw
    │   └── trimmed
    ├── references
    │   ├── annotation
    │   ├── genome
    │   ├── sra_info
    │   └── star_index
    ├── results
    │   ├── alignments
    │   ├── counts
    │   └── qc
    └── scripts
  ```
- Retrieve the data from GSE72548 using `esearch` and `fasterq`, similar to what we had done in M5L2.
- Run `fastp` on the data set to trim and assess the quality of the reads
- Build a `STAR` index using the TAIR10 Genome and annotation from Ensembl
- Run `STAR` on the 24 samples
- Use `qualimap` to assess the alignment, and assemble a `multiqc` report
- Use `featureCounts` to generate counts for the alignments
- Explore the count data, and assess the quality of the counts (look at clustering, and dispersion estimates to see if the samples work well)
- Run DESeq2 on the samples using a 2x2 factorial design, similar to what we had done this week in M6P2, and also in M4P1
- Perform a functional analysis to investigate the role of top DEGs across treatment and genotype, identifying candidates that could play a role in the cytokinin signaling pathway, thought to be responsible for pathogen response.*(If your results disagree with the paper, how? and why do you think differently)*
- Reference literature (The reference paper is acceptable)

## Hints, Tips, and Tricks: 

1. If you use Jupyter Notebook, use the bash terminal and the environment `Module_5_STAR`. Use the Hellbender Desktop to run the bash commands, or submit the commands as a bash script to the SLURM scheduler via SSH - [Hellbender Documentation](https://docs.itrss.umsystem.edu/pub/hpc/hellbender)
2. Use the `STAR` module, rather than the version within the `Module_5_STAR` environment, for use in alignment, as you can leave the --tmpOutDir blank and the cluster will manage it for you, avoiding issues with multiple users writing to `/tmp`. 
3. Document EVERY SINGLE step - this will help you if you stop in the middle of the pipeline
4. Use R-Studio for the analysis, either in the Hellbender Desktop or locally. You can download your count matrices and metadata via [Globus](https://www.globus.org/) or use `scp` via SSH terminal.
5. If you use Jupyter Notebook for the analysis, use the `Module 3 - R` kernel, and follow along from the DESeq analysis you completed in M4P1, M4P2, M4E1, and M6P2.
6. We will hold office hours for help this week. If you are having trouble, reach out via Canvas or email.
