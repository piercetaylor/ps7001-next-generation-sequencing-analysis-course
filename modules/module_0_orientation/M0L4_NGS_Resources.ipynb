# Next Generation Sequencing Techniques Resources
These videos, links, and notebooks provide an overview of the three NGS techniques that will be covered in this course: RNA-Seq, CHiP-Seq, and DAP-seq.
It is essential to recognize that various approaches exist for pre-processing and visualizing these workflows, making it crucial to understand your goals when applying these tools to your own datasets.
