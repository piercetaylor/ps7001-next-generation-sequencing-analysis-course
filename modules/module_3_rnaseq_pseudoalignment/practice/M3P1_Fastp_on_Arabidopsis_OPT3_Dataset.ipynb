   - HINT: You can use a fastp command similar to this:

```
for file in data/raw/*.fastq.gz
do
  base=$(basename "$file" .fastq.gz)
  fastp \
    -i "$file" \
    -o "data/trimmed/${base}_trimmed.fastq.gz" \
    --html "results/qc_trimmed/${base}_fastp.html" \
    --json "results/qc_trimmed/${base}_fastp.json" \
    --thread 4 \
    --detect_adapter_for_pe \
    --trim_front1 5 \
    --cut_tail \
    --length_required 30
done
```
