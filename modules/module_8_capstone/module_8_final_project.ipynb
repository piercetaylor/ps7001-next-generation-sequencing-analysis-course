# Module 8 Final Project

### **SAVE ALL WORK FOR YOUR PROJECT WITHIN THE MODULE_8 FOLDER**
For the final project, apply a pipeline from the course to a real-world dataset:

- RNA-Seq using pseudo-alignment to the transcriptome with Salmon
- RNA-Seq using genome (or transcriptome) alignment with STAR
    - For genome browser visualization: Generate alignment and coverage files for visual inspection in a genome browser
    - For the generation of count matrices and subsequent DE analysis and functional annotation
    - For Isoform quantification, SNP/Indel calling, Allele-Specific Expression (ASE) - See resource links if you are interested

### Dataset Selection:

The data set you choose is up to you! It can be a publicly available dataset, either from a database or a published paper, or you could use your own data if you have it available. The data set must have a clear experimental design, multiple treatments, and at least 3 biological replicates per treatment group. Sequencing and sample information must be well documented. You must perform a full analysis from raw sequencing data to final biological interpretation.

 

To find a public dataset, I recommend searching the NCIB GEO database

Links to an external site. or finding papers relevant to the course and your research.

### Requirements:

1. ) Project Report (1-2 pages for undergraduate students and 3-5 pages for graduate students)

    Introduction: Describe your biological question and experimental design, and how your data is applicable. Why is your chosen pipeline appropriate, and what are you hoping to learn in this analysis? (Are you copying a tutorial, or is this new information?)
    Methods: Provide clear instructions to your pipeline: tools used, parameters, versions, and scripts
    Results: Present the findings of your analysis, including at least 2 figures and an interpretation of these results in reference to your experiment
    Discussion/Conclusion: Discuss findings, limitations of your analysis, and suggestions for future directions

2. ) Scripts or Jupyter Notebooks used:

    Provide the code you used, either in Jupyter Notebook or via Bash + R scripts, in an organized format. You should follow the project organization we talked about in class. Separate your data, results, references, environments, and scripts neatly in a way that is reproducible. Scripts must be clearly commented and organized.

3. ) Files Generated for output:

    It is not feasible for you all to submit the data sets you used, but the submission of count matrices, DEG lists, and NarrowPeak or BigWig tracks will be required. These files, along with your scripts and final report, should be submitted to your student folder within the class directory. Your Final Report will also be submitted to Canvas in Module 8.
