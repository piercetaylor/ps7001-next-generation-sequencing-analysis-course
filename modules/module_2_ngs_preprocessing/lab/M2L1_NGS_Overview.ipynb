## NGS Sequencing Review

Illumina sequencing follows these steps:
1. Fragmenting DNA/RNA into small pieces
2. Ligating adapters to fragments
3. Amplifying fragments on a flow cell (bridge amplification)
4. Sequencing by synthesis using fluorescently labeled nucleotides

### Inputs required for Illumina Sequencing:
- DNA or RNA extracted from samples
- Library preparation kit (for adapters)
- Sequencing primers
- Flow cell and sequencing reagents

Read more about [Illumina Sequencing](https://www.illumina.com/content/dam/illumina-marketing/documents/products/illumina_sequencing_introduction.pdf)

## RNA-Seq Pipeline Overview
*[Third-party RNA-seq workflow figure omitted from the public version. A similar figure is in the [HBC Intro-to-DGE lessons](https://hbctraining.github.io/Intro-to-DGE/).]*


RNA-Seq is used to quantify gene expression. It measures the quantity and sequences of RNA within a sample, providing a snapshot of gene expression. RNA-Seq is used to identify which genes are activated, quantify gene expression, and detect alternative splicing or transcript variants. In the lab, RNA-Seq is used to compare gene expression across treatments, tissues, and time points to understand gene regulation within the samples.

The main steps involved in the processing of RNA-Seq Data are:

### Step 1: **Pre-processing**
- Inputs: Raw sequencing reads (`fastq` files, usually paired-end)
- Quality check (`FastQC`)
- Adapter trimming (`Trimmomatic`)
- Filtering low-quality reads (Reads that are too short, have high ambiguity, etc.)
- Outputs: Trimmed and QC checked `fastq` files

### Step 2: **Alignment / Mapping**
- Inputs: Trimmed and QC checked `fastq` files, Reference Genome/Transcriptome `fasta` file
- Align reads to a reference genome or transcriptome (Genome alignment: `HISAT2`, `STAR` / Pseudo-Alignment: `Salmon`, `Kallisto`)
- Outputs: `BAM` files containing aligned reads
- Example tools: `bowtie2`, `hisat2`, `STAR`

### Step 3: **Annotation**
- Inputs: Aligned reads in `BAM` file format, gene annotation file (`GTF`/`GFF3`)
- Assign reads to genes/transcripts
- Generate count matrices for downstream analysis
- Output: Counts, or normalized counts (TPM or FKPM) are produced in `CSV`, `TSV`, or `TXT` format. 
- Example tools: `featureCounts`, `HTSeq-count`\
  

### Step 4: **Post-Processing**

After alignment and annotation, post-processing allows researchers to interpret RNA-Seq data biologically. It typically includes the following components:

- **Inputs**
    - Count matrix (either raw or normalized into TPM or FKPM)
    - Sample metadata (treatments, conditions, replicates, sequencing lanes, etc.)

- **Quality Assessment**
    - Verify that the alignment and annotation steps were successful.
    - Check for potential biases, batch effects, or sample outliers.
    - Tools like `MultiQC` can summarize quality metrics across samples.

- **Visualization**
    - PCA (Principal Component Analysis): reduces dimensionality to visualize sample relationships and batch effects.
    - Volcano plots: display differential expression, showing significance versus fold-change for genes.

- **Differential Expression Analysis** (`DESeq2`, `edgeR`)
    - Identify genes whose expression changes significantly between conditions or treatments.
    - Outputs include: List of DE genes in log fold-change, p-values, and adjusted p-values for multiple testing.
    - Heatmaps: *Done after DE generally* Visualize expression levels of selected genes across samples, highlighting patterns and clusters.

- **Functional Enrichment Analysis**
    - Determine if sets of DE genes are overrepresented in particular biological pathways or processes.
    - Gene Ontology (GO) analysis identifies biological processes, molecular functions, and cellular components in reference to known genes.
    - KEGG pathway analysis highlights affected metabolic or signaling pathways.





## CHiP-Seq

CHiP-Seq is used to identify DNA regions within the genome bound by transcription factors, allowing researchers to identify potential regulatory targets within the genome. CHiP seq is expensive, time-consuming, and occurs in vivo. 

Steps in CHiP-Seq:
- Crosslink proteins to DNA
- Immunoprecipitate DNA-protein complexes
- Sequence the pulled-down DNA fragments
- Map reads to the genome
- Identify peaks (protein binding sites)
- Visualization in a genome browser and downstream analysis (IGV or UCSC Genome Browser)

## DAP-Seq

DAP-Seq (DNA Affinity Purification Sequencing) identifies binding sites of transcription factors in vitro. Similar to CHiP-Seq in the way it is used to profile genome-wide binding of transcription factors. This technique is high-throughput and requires less time than CHiP-Seq, but it does not account for the cellular context of TF-DNA binding (chromatin structure, cofactors, etc.)

Steps in DAP-Seq:
- Express and purify the TF of interest
- Incubate TF with genomic DNA fragments
- Sequence the bound DNA fragments
- Map reads to the genome
- Identify peaks and analyze target genes

## Summary

| Pipeline  | Purpose | Main Steps |
|-----------|---------|------------|
| RNA-Seq   | Gene expression profiling | QC → Trim → Align → Annotate → DE Analysis |
| ChIP-Seq  | Identify protein-DNA interactions | Crosslink → IP → Sequence → Align → Peak Call |
| DAP-Seq   | In vitro TF binding sites | TF purification → Bind DNA → Sequence → Align → Peak Call |
