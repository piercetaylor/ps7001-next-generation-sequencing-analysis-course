```
>AT1G01010.1 | N-acetyltransferase
ATGGCTTCTTCTTCTTCTTCTGGTGCTGCTGCTGCTGCTG
>AT1G01020.1 | Hypothetical protein
ATGGCGGCGGCGGCGGCGGCGGCGGCGGCGGCGGCGGCGGC
>AT1G01030.1 | Ribosomal protein L2
ATGGCAGGAGGAGGAGGAGGAGGAGGAGGAGGAGGAGGAGGA
```

FASTA's are formatted as a `.fa`, `.fasta`, or `.fna` files. 

### FASTQ files

- A FASTQ file is used to store raw sequencing reads from high-throughput sequencing techniques, like Illumina. Unlike FASTA files, FASTQ files store quality information about every base read by the sequencing machine. In the case of Illumina sequencing, a quality score is assigned from the cluster of the fragment that is imaged base by base in a flow cell.
- Each FASTQ file has the following format:

<u>Header line:</u> starts with @ followed by a read identifier (and sometimes additional info like the sequencing run or flow cell coordinates).

<u>Sequence line:</u> the DNA or cDNA sequence of the read.

<u>Plus line:</u> a + symbol, sometimes repeated with the read identifier (mostly a separator).

<u>Quality line:</u> ASCII-encoded *Phred quality* scores indicating the confidence of each base call (higher scores = higher confidence).

A FASTQ file would look like this: 

```
@AT1G01010_read1
ATGGCTTCTTCTTCTTCTTCTGGTGCTGCTGCTGCTGCTG
+
IIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIII
@AT1G01020_read2
ATGGCGGCGGCGGCGGCGGCGGCGGCGGCGGCGGCGGCGGC
+
IIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIII
@AT1G01030_read3
ATGGCAGGAGGAGGAGGAGGAGGAGGAGGAGGAGGAGGAGGA
+
IIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIII
```

FASTQ's are formatted as `.fastq`, `.fq`, or `.fastq.gz` file types.

NOTE: The `.gz` extension is always assigned to a file that has been `gzipped` and is a compressed file format. This is common to save space and will need to be extracted to be read properly. Many programs and tools are equipped to handle `.gz` files natively, but Windows computers will not display the `.gz` file extension within the File Browser, and this can cause confusion. 

## GTF/GFF3 files

GTF (Gene Transfer Format) and GFF (General Feature Format) are text-based formats used to describe genomic features on a reference genome. These features include genes, transcripts, exons, coding sequences (CDS), UTRs, and regulatory elements.

- GFF is the broader standard (GFF3 is the latest version).

- GTF is a slightly stricter variant commonly used in RNA-seq pipelines.

- These files serve as annotations, helping aligners and analysis tools know where genes and exons are located.

Each line in a GTF/GFF file represents a feature and contains multiple tab-separated fields.


    seqname – chromosome or scaffold name (e.g., Chr1)
    
    source – software or database that generated the annotation (e.g., TAIR10)
    
    feature – type of feature (gene, transcript, exon, CDS)
    
    start – start coordinate on the chromosome
    
    end – end coordinate on the chromosome
    
    score – optional numeric score (often . if not used)
    
    strand – + or - indicating the DNA strand
    
    frame – reading frame for coding sequences (0, 1, 2), . if not applicable
    
    attribute – key-value pairs, such as gene_id, transcript_id, gene_name

An example GTF file is shown below
```
Chr1  TAIR10  gene      3631   5899   .  +  .  gene_id "AT1G01010"; gene_name "NAC1";
Chr1  TAIR10  transcript 3631   5899   .  +  .  gene_id "AT1G01010"; transcript_id "AT1G01010.1";
Chr1  TAIR10  exon      3631   3913   .  +  .  gene_id "AT1G01010"; transcript_id "AT1G01010.1";
Chr1  TAIR10  CDS       3679   3913   .  +  0  gene_id "AT1G01010"; transcript_id "AT1G01010.1";
```

## SAM and BAM files
A SAM file is a text-based format that stores sequencing reads aligned to a reference genome. SAM files are the readable precursors to BAM files, which are their compressed binary form.
- SAM files are generated by aligners such as Bowtie, BWA, HISAT2, or STAR.

A SAM file has two main parts:

1. Header lines: start with @ and provide metadata, e.g., reference genome names, sequence lengths, and aligner information.
```
@HD VN:1.6 SO:coordinate
@SQ SN:Chr1 LN:30427671
@SQ SN:Chr2 LN:19698289
```
2. Alignment records (one per read), tab-delimited fields displayed in the table below:
| Field           | Description                                                                |
| --------------- | -------------------------------------------------------------------------- |
| QNAME           | Read identifier (from FASTQ)                                               |
| FLAG            | Bitwise flag indicating mapping properties (strand, paired-end, duplicate) |
| RNAME           | Reference sequence name (chromosome)                                       |
| POS             | 1-based leftmost mapping position                                          |
| MAPQ            | Mapping quality (confidence)                                               |
| CIGAR           | Alignment operations (matches, insertions, deletions)                      |
| RNEXT           | Mate reference name (for paired-end reads)                                 |
| PNEXT           | Mate position                                                              |
| TLEN            | Template length                                                            |
| SEQ             | Read sequence                                                              |
| QUAL            | Base quality scores (ASCII)                                                |
| Optional fields | Additional info (e.g., alignment score, number of mismatches)              |


Every line (after the header) corresponds to one read and its alignment. It is important to understand the layout of these files in case they need to be parsed when searching for a specific transcript. 


BAM files, or Binary Alignment Map files, are binary compressed files used to store aligned sequencing reads. BAM files are the binary version of SAM files and are compressed. This format utilizes random access, which makes for efficient viewing or analysis of specific regions of a genome. 

Although BAM is binary, conceptually it contains the same information as a `SAM` file.
