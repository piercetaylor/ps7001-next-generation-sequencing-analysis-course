## Hellbender Desktop

The hellbender desktop will be used throughout the rest of this course. It is a way to access the Hellbender HPC system and run a terminal shell similar to Jupyter Notebook. The hellbender desktop uses [NOVNC](https://novnc.com/info.html), an open-source virtual network computing (VNC) client that allows users to connect to a virtual desktop. The NOVNC client is web-based, meaning that it can operate from a browser on any operating system and runs fully inside a browser window. This virtual desktop offers remote access to a graphical user interface (GUI) for the distribution of Alma Linux running on Hellbender.


## Launching a Hellbender Virtual Desktop

#SHIFT + ENTER TO SEE VIDEO
<iframe width="560" height="315" src="https://www.youtube.com/embed/sIWVIYC0JmE?si=zOMNpxc1CLwktPKF" title="YouTube video player" frameborder="0" allow="accelerometer; autoplay; clipboard-write; encrypted-media; gyroscope; picture-in-picture; web-share" referrerpolicy="strict-origin-when-cross-origin" allowfullscreen></iframe>

To launch a session for Hellbender Virtual desktop visit [RSSOnDemand](https://hb-classes.missouri.edu). Select the Hellbender Desktop. 

![screen](../resources/images/M2L2HellbenderDesktopOnDemand.png)


Then, adjust the needed resources for the session. 

![screen](../resources/images/M2L2HellbenderDesktopConfig.jpg)


## Running a Terminal
A terminal session can be opened from the Hellbender Desktop VNC client, and it will function similarly to the terminal sessions created in Jupyter Notebook. Please watch the video below. 

#SHIFT + ENTER TO SEE VIDEO
<iframe width="560" height="315" src="https://www.youtube.com/embed/JyP1WOZS1_8?si=zagPNm01GOn2bC4F" title="YouTube video player" frameborder="0" allow="accelerometer; autoplay; clipboard-write; encrypted-media; gyroscope; picture-in-picture; web-share" referrerpolicy="strict-origin-when-cross-origin" allowfullscreen></iframe>

## Using the File Browser
The File Browser application operates within Hellbender Desktop's VNC connection and functions similarly to the File Explorer in Windows or the Finder application in macOS. The File Browser is an application that enables users to access files within the system through a Graphical User Interface (GUI). 

**On the Hellbender Desktop VNC Client, the home folder is the storage folder associated with your RSS Account `$HOME/`. This home folder is where your 50 GB of personal storage is allocated. To navigate to your student folder within the course, you have two options:** 
1. Navigate to the symlink you set up in Module 1 Lab 2 called `MyStudentFolder`. If you do not have a symlink setup, you can navigate to your home directory with `cd`, then run the command `ln -s $WORK_DIR/ ./MyStudentFolder`, replacing `{$PawPrint}` with your 5-digit pawprint. After this, you will see a symlink for `MyStudentFolder` within the Hellbender Desktop's File Browser application.

2. You can navigate to the course directory manually within the terminal, using the command `cd $WORK_DIR` where `{$PawPrint}` is replaced by your 5-digit pawprint. 

#SHIFT + ENTER TO SEE VIDEO
<iframe width="560" height="315" src="https://www.youtube.com/embed/4Pt_iJhMhkA?si=5CyRIzV_0OS5Xvb0" title="YouTube video player" frameborder="0" allow="accelerometer; autoplay; clipboard-write; encrypted-media; gyroscope; picture-in-picture; web-share" referrerpolicy="strict-origin-when-cross-origin" allowfullscreen></iframe>

## Why use the Desktop?

The Hellbender Desktop provides a graphical user interface (GUI) through the VNC client. This allows students to browse the web within the same screen they have the terminal in, looking for answers from [Stack Overflow](https://www.stackoverflow.com) or using AI to assist in determining commands, writing bash scripts, or general issues they may encounter when running commands in the processing of NGS data. 

Students must understand how to ask the right questions about their code and terminal usage so that the answers they receive will apply to their use case. The result of copying and pasting code into AI often leads to programs that are non-functional or produce the wrong output. For more information on how to use Generative-AI effectively in coding and terminal usage, see [M0L2_Prompting_For_Success](../../module_0_orientation/M0L2_Prompting_For_Success.ipynb)
