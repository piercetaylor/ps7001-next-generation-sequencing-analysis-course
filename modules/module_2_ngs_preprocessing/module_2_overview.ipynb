# Module 2 Checklist
## Module 2 - File Formats and NGS Techniques
This module provides an overview of NGS techniques, including Illumina Sequencing, RNA-Seq, CHiP-Seq, and DAP-Seq. This week, we will also practice command-line skills and the pre-processing of RNA-Seq data in preparation for pseudo-alignment next week. 

## Module Topics: 

-    Illumina Sequencing
-    Overview of RNA-Seq data processing
-    Introduction to CHiP-Seq and DAP-Seq concepts
-    File types used in NGS data analytics
-    File management and parsing within bash: `samtools`, `awk`, `grep` 
-    Pre-processing of RNA-Seq data: quality control checks and trimming
