# Module 8 - Optional scRNA-Seq Lab and Practice

### **This is an optional lab and practice lesson covering the basics of scRNA-Seq, and analysis in Seurat**
For the final project, students may use a scRNA-seq data set of their choice as well. In this case, the following materials are provided to allow for students to run this type of analysis, or for learning purposes. This week's labs and practices will not affect the student's final grade.

**NOTE: M8_L2 downloads about 1.5 GB of FASTQ files and runs STAR, so create the Jupyter session with CPU cores = 8 and RAM (memory) >= 32 GB, and allow 1 to 2 hours. M8_P1 runs in R and needs about 8 GB of RAM (16 GB for the optional mutant section).**

## Module Topics:
- How single-cell RNA-Seq differs from bulk RNA-Seq
- Protoplasts and nuclei: preparing plant cells for scRNA-Seq
- 10x Genomics Chromium library preparation: GEMs, cell barcodes, UMIs and 3' capture
- The structure of 10x FASTQ files (R1, R2, I1) and how they are QC'd
- Alignment, barcode correction, UMI counting and cell calling with STARsolo
- Cell QC, normalization, integration, clustering and marker genes in Seurat v5
- Annotating *Arabidopsis* root cell types with known marker genes
