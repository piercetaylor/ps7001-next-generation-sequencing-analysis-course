# What is Hellbender HPC?

Hellbender is the latest High-Performance Computing (HPC) resource available to researchers and students (with sponsorship by a principal investigator) within the UM System.

Hellbender consists of 208 mixed x86-64 CPU nodes (112 AMD, 96 Intel), providing 18,688 cores as well as 28 GPU nodes consisting of a mix of Nvidia GPUs (see hardware section for more details). Hellbender is attached to our Research Data Ecosystem (RDE) that consists of 8PB of high-performance and general-purpose research storage. RDE can be accessed from other devices outside of Hellbender to create a single research data location across different computational environments. 

Hellbender's operating system is [AlmaLinux 8.10](https://almalinux.org/). This enables easy operation of RHEL-compatible software, ensures secure system operation, and allows for a high degree of scalability within the system. On the User's end, Hellbender acts as a remote server for processing computationally intensive data sets, such as large-scale genomics projects or complex physics problems. Users can connect to the server to run their programs or jobs through scheduled or interactive sessions. Hellbender is accessible through [OnDemand](https://ondemand.rnet.missouri.edu) for use of the VSCode server, a Virtual Desktop, Jupyter Notebook, etc. A direct SSH connection via the SSH client of your choice is also available and convenient for transferring files via SCP from Windows Machines without the need for Globus. 

### Hellbender data storage
Each user is granted 50gb of space in their home folder `$HOME`. In this course, data will be stored mainly within your course folder `$WORK_DIR`. Global environments will be utilized for later modules in the course and will be stored within the course folder. 

Researchers who utilize the HPC cluster are also granted 500 GB of storage in their data partition `$HOME/data`. It is recommended that data used in your research, such as Conda environments, working data, and large files, be stored in the data partition.

**If you are not enrolled as a researcher with ITRSS and you would like to gain access to this storage, please visit the [Researcher Account Request Form](https://missouri.qualtrics.com/jfe/form/SV_e4E2UyI77SpiYMC) from ITRSS.**

# Terminals and Command Lines

The Linux terminal, A.K.A. the command line, is a tool for managing files, running programs, and in our case, processing large biological datasets. Instead of using a graphical user interface (GUI), like the one found on a Windows or Mac computer, to browse files or launch programs, commands are input directly into the terminal. Bash, or Bourne Again SHell, is the programming language used by the command line. Bash commands can also be written in a script to quickly iterate through a series of repetitive commands, saving time and ensuring precision. Bash scripts are generally written in the .sh file extension, and we will be utilizing them in several of the modules in this course.
