# Module 1: Introduction to Genomics

This module is an introduction to command-line bioinformatics skills, including navigating Linux environments, scripting, and environment creation with Conda. 

## Module Topics: 
1. Introduction to NGS Techniques and Course Objectives
2. Configuration of Jupyter Notebook, R, Python, and Hellbender HPC
3. Introduction to bash, basic scripting, and command line navigation
4. Creation and management of Conda environments within the cluster
