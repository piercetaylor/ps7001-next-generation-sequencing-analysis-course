# <u>What is a Bash Script<u>

In [M1L2](../lab/M1L2_Introduction_To_Bash_Via_Hellbender.ipynb), we learned about utilizing bash commands within the terminal of Jupyter Notebook (which is a direct connection to the HPC node you are utilizing to host Jupyter). Bash is a dynamically typed scripting language used to control a computer system, such as an HPC node, through command-line instructions. While it's not a general-purpose programming language like Python or R, Bash is well-suited for tasks such as moving files, running software tools, and managing system processes. 

A Bash script is a plain text file with the file extension `.sh` that contains a series of bash commands. These scripts are used to automate tasks that you would normally type one by one into a command line terminal. In this course, we will use Bash scripting to execute multiple commands in sequence—an essential skill for running tools commonly used in next-generation sequencing (NGS) workflows. 

For example, if you wanted to run `fastqc` (a quality control tool for `.fastq` files) on 100 samples, you could write a Bash script that loops through all the filenames and runs `fastqc ${FileName}.fastq` for each one. This allows the HPC node to process all 100 files with a single script, saving you from having to type the command 100 times manually. While the total compute time remains the same, the scripting approach dramatically reduces manual effort and ensures consistency.

# <u>Data, Variables, STD IO<u>
Bash is not a strongly typed language, and it treats most data as strings by default. However, it effectively supports three basic data types:

* `Integer` – Whole numbers (e.g., `5`, `42`)
* `Float` – Decimal numbers (Bash does not handle floats natively, but they can be processed using tools like `bc`)
* `String` – Any text or sequence of characters

### Standard Output
In Bash, standard output refers to text or data that is printed to the terminal. There are two commonly used commands to write output:
* `echo` – Used to print literal text or the value of variables.
* `printf` – Used for more controlled and formatted output.

**Using `echo`**

The `echo` command is simple and flexible. It can be used to display static text or variables:

````python
%%bash 
varName="value"
echo "This is a message"
echo "The value of varName is $varName"
````

**Using `printf`**

The `printf` command provides more precise control over formatting and layout. Its syntax is: `printf <format string> <space-separated arguments>`. These specifiers tell printf how to interpret and display each argument you provide.

Each format specifier begins with a percent sign (%) and is followed by a character that defines the data type and how it should be displayed.

| Specifier    | Description                                | Example Usage                | Output Example   |
| ------------ | ------------------------------------------ | ---------------------------- | ---------------- |
| `%s`         | String                                     | `printf "%s\n" "Hello"`      | `Hello`          |
| `%d` or `%i` | Signed decimal integer                     | `printf "%d\n" 42`           | `42`             |
| `%u`         | Unsigned decimal integer                   | `printf "%u\n" 300`          | `300`            |
| `%f`         | Floating-point number (default 6 decimals) | `printf "%.2f\n" 3.14159`    | `3.14`           |
| `%e`         | Scientific notation (lowercase `e`)        | `printf "%e\n" 12345.6789`   | `1.234568e+04`   |
| `%E`         | Scientific notation (uppercase `E`)        | `printf "%E\n" 12345.6789`   | `1.234568E+04`   |
| `%g`         | Use `%f` or `%e` depending on value        | `printf "%g\n" 0.00012345`   | `0.00012345`     |
| `%x`         | Hexadecimal (lowercase)                    | `printf "%x\n" 255`          | `ff`             |
| `%X`         | Hexadecimal (uppercase)                    | `printf "%X\n" 255`          | `FF`             |
| `%o`         | Octal                                      | `printf "%o\n" 8`            | `10`             |
| `%%`         | Literal percent sign                       | `printf "Progress: 100%%\n"` | `Progress: 100%` |

# <u>Basic Operators<u>
In Bash, you can perform arithmetic calculations using the syntax `$(({$Expression}))`
Here, `{$Expression}` is any valid arithmetic expression.

By default, numbers are interpreted as decimal (base 10) unless specified otherwise:

* Decimal (base 10): Numbers without any special prefix, e.g., `42`
* Octal (base 8): Numbers with a leading zero, e.g., `075` (which equals 61 in decimal)
* Hexadecimal (base 16): Numbers prefixed with `0x`, e.g., `0x1A` (which equals 26 in decimal)


| Operator      | Description                 | Example                     | Explanation                                |
| ------------- | --------------------------- | --------------------------- | ------------------------------------------ |
| `+`           | Addition                    | `echo $((3 + 4))`           | Outputs `7` (3 plus 4)                     |
| `-`           | Subtraction                 | `echo $((10 - 6))`          | Outputs `4` (10 minus 6)                   |
| `*`           | Multiplication              | `echo $((5 * 3))`           | Outputs `15` (5 times 3)                   |
| `/`           | Integer division            | `echo $((10 / 3))`          | Outputs `3` (integer division)             |
| `**`          | Exponentiation              | `echo $((2 ** 4))`          | Outputs `16` (2 to the power of 4)         |
| `%`           | Modulo (remainder)          | `echo $((10 % 3))`          | Outputs `1` (remainder of 10 divided by 3) |
| `-=`          | Subtraction assignment      | `let "a=10; a-=3; echo $a"` | Outputs `7` (a = a - 3)                    |
| `+=`          | Addition assignment         | `let "a=5; a+=4; echo $a"`  | Outputs `9` (a = a + 4)                    |
| `*=`          | Multiplication assignment   | `let "a=3; a*=6; echo $a"`  | Outputs `18` (a = a \* 6)                  |
| `/=`          | Integer division assignment | `let "a=20; a/=5; echo $a"` | Outputs `4` (a = a / 5)                    |
| `%=`          | Modulo assignment           | `let "a=11; a%=4; echo $a"` | Outputs `3` (a = a % 4)                    |
| `parameter++` | Post-increment              | `let "a=5; echo $((a++))"`  | Outputs `5` then increments `a` to 6       |
| `parameter--` | Post-decrement              | `let "a=5; echo $((a--))"`  | Outputs `5` then decrements `a` to 4       |
| `++parameter` | Pre-increment               | `let "a=5; echo $((++a))"`  | Increments `a` to 6 then outputs `6`       |
| `--parameter` | Pre-decrement               | `let "a=5; echo $((--a))"`  | Decrements `a` to 4 then outputs `4`       |


````python
%%bash 
echo $((10 + 5))       # Outputs 15 (decimal addition)
echo $((0xA + 5))      # Outputs 15 (hexadecimal 0xA = decimal 10)
echo $((2#1010 + 5))   # Outputs 15 (binary 1010 = decimal 10)
````

# <u>Conditionals and Branching<u>

Conditionals—also known as logical statements—allow a script to make decisions and control the flow of execution based on certain conditions. These are essential for creating flexible and responsive Bash scripts.

We use `if` statements in Bash to control the flow of a script based on conditions. Suppose the condition evaluates to true (i.e., returns an exit status of 0), the commands inside the `if` block are executed. If the condition is false (i.e., a non-zero exit status), the script skips the `if` block and either proceeds to the next part of the script or executes the `else` block, if present. The command `fi` signals to the computer that the bash script is complete.

###  **One-Way Selection**

The simplest form of a conditional uses if to test whether a command executes successfully using an `if` statement:

```
if a_command; then
    command1
    command2
fi
```

If `a_command` exits with status 0 (meaning success), then command1 and command2 are executed.
If `a_command` fails (non-zero exit status), the script skips the commands inside the if block.

###  **Exit Status**

Every command in Bash returns an exit status, which is an integer:

* `0` → Success
* Non-zero (e.g., `1`, `2`, `127`) → Failure or error

The special variable `$?` stores the exit status of the most recently executed command.

````python
%%bash
ls /nonexistentfolder
echo $? # Outputs a non-zero value, indicating failure
````

### **Two-Way Selection**
Allows a script to choose between two actions based on a condition using `if` and `else`.
```
if [ condition ]; then
    # commands if the condition is true
else
    # commands if the condition is false
fi
```

````python
%%bash
#!/bin/bash
a=100
b=20
if (( $a < $b )); then
    echo "$a is less than $b"
else
    echo "$a is not less than $b"
fi
exit 0
````

# <u>Loops and Arrays<u>

### **Loops**
Loops allow us to repeat a set of commands multiple times, just like repetitive tasks in real life (e.g., slicing a carrot).
Slicing a carrot involves a loop:
1. Get a cutting board
2. Get a knife
3. Place the carrot on board
4. Lift knife
5. Advance carrot
6. Slice the carrot
7. If the carrot is not done, repeat from step 4.
8. If the carrot is finished, stop.
    * Steps 4–7 form a loop: repeat until a condition is met!

### There are two main types of loops in Bash. A `for` loop and a `while` loop.

A `for` loop is used when: 
* You know in advance how many times you want to loop.
* You're looping through a list or a range.

```
for (( init; condition; update )); do 
    # body
done

for val in val1 val2 ...; do
    # body
done
```

There are two types of `for` loops: 

C-style `for` loop (suitable for numbers)

````python
%%bash
# Start a for loop where:
# i starts at 1
# loop continues while i is less than or equal to 5
# i increments by 1 after each loop iteration
for (( i=1; i<=5; i++ )); do
    # Print the current iteration number
    echo "Iteration $i"
done
````

List-style `for` loop (good for items in a list)

````python
%%bash
# Loop over a list of names: Alice, Bob, and Charlie
for name in Alice Bob Charlie; do
    # Print a greeting for each name
    echo "Hello, $name!"
done
````

For loops are very handy, and can be used to list, modify or name files. 

````python
%%bash
# print line and total characters in each line for contents of home dir
count=1
for i in $(ls ~); do
    echo "Line $count ($i) contains $(echo -n $i | wc -c) characters"
    (( count++ ))
done
````

A `while` loop is used when: 
* You want to repeat something as long as a condition is true. This is useful when you don’t know in advance how many times you want your loop to repeat.
```
while [ condition ]; do
    # loop body
done
```

````python
%%bash
count=1
while (( $count <= 20 )); do
    if (( $count%2 == 0 )); then
        echo $count
        #exit 0;
    fi
    (( count++ ))
done
````

### **Arrays**
Arrays allow you to store multiple values under a single variable name. Bash supports one-dimensional arrays only. Arrays are indexed using integers, with the default starting index being `0`. This means that if you have an array called `arr` with the values `arr=("apple", "banana", "orange")` then `arr[0]` would return `"apple"` and `arr[1]` would return `"banana"`.


Key Characteristics:

1. Elements are accessed using indices.

2. Bash arrays behave somewhat like Python dictionaries or C++ maps:

    * Indices act like keys.

    * Do not have to start at 0.

    * Do not need to be contiguous.

### Declaring Arrays
There are several ways to declare arrays:
1. Using the declare command: 
   `declare -a arrName`
2. Assigning a value to a specific index: 
`arrName[2]="value"`
3. Declaring and initializing(assinging value) in one step:
`arrName=("value1" "value2" "value3")`
   * Bash automatically assigns indices starting from 0 in this case.

````python
%%bash
#Declare an array called names
declare -a names

# assigning values
names[0]=John
names[1]="Jane Smith"
names[2]="Michael"

## access elements
echo ${names[*]} # access all elements
echo ${names[0]} # access element at index 0
echo ${names[1]}
````

````python
%%bash
# declare just like variables
days[0]=Sunday
days[1]=Monday

# access elements
echo ${days[*]} # can use @ as well!
echo ${days[0]}
echo ${days[1]}

# appending elements to the end
days+=(Tuesday Wednesday)
echo ${days[*]}

# deleting element
unset days[0] #drops the first element, "Sunday"
echo ${days[*]}
````

### Looping through arrays
You can loop through arrays in two main ways:
1. Loop through values:

````python
%%bash
arr=("apple" "banana" "cherry")

for val in "${arr[@]}"; do
    echo "$val"
done
````

2. Looping through indices: 
* **Note:** `[@]` expands to all elements in the array.

````python
%%bash
arr=("apple" "banana" "cherry")

for val in "${arr[@]}"; do
    echo "$val"
done
````

````python
%%bash

# Populate an array 'nums' with squares of numbers from 1 to 20
for i in {1..20}; do
    nums[$i]=$(( i**2 ))  # Store i squared in the array
done

# Print all values in the array using a value-based loop
for i in ${nums[*]}; do
    printf "%d " "$i"  # Print each value followed by a space
done

printf "\n"

# Get the total number of elements in the array
len=${#nums[*]}  # This gives the number of assigned indices

echo "nums length = $len"

# Use a while loop to iterate through the array by index
i=1
while (( i <= len )); do
    printf "%d " "${nums[i]}"  # Print the i-th element
    (( i++ ))  # Increment index
done

printf "\n"

# Demonstrate handling sparse arrays
unset nums[10]  # Remove the 10th element
unset nums[20]  # Remove the 20th element

# Print all defined values using their indices
# ${!nums[*]} expands to defined indices only
for i in ${!nums[*]}; do
    printf "%d " "${nums[i]}"
done
````

Viewing the entire file is somewhat disorganized. If we only wanted to see a portion of the fasta file, say the first six lines, we would add the pipe operator `|` -  meaning It takes the output of the command on the left side and uses it as the input for the command on the right side - and the command `head` to display only the first 10 lines. 

````python
%%bash
while read line; do
    echo "$line"
done < ~/test/ArabidopsisCDNA.fa | head
````

Now, say that we only wanted to print the headers of the `fasta` file. We can use the `grep` command within our bash scripts, similar to how we used it within the [M1L2](../lab/M1L2_Introduction_To_Bash_Via_Hellbender.ipynb). 

````python
%%bash

# Use grep to extract lines starting with ">", which indicate FASTA headers
grep "^>" ~/test/ArabidopsisCDNA.fa | 

# For each matching header line, run the following while-loop
while read header; do
    # Print a message indicating a sequence header was found
    echo "$header"
done
````


### File Output
Similarly, we can output all the Sequence IDs, the first word behind the `>`, to a text file. To export to a file within your bash script, you use redirection operators, such as `>`. After creating this new file, we will use a while loop to read the Sequence IDs from the new file one by one. 

````python
%%bash
# Step 1: Extract sequence IDs and write to a new file
grep "^>" ~/test/ArabidopsisCDNA.fa | cut -d ' ' -f1 | sed 's/^>//' > ~/test/ArabidopsisCDNA_ids.txt

# Step 2: Read the new file line by line
#we will use variable i to count the number of Sequence IDs
i=1
while read id; do
    echo "Sequence ID $i: $id"
    ((i++))
done < ~/test/ArabidopsisCDNA_ids.txt
````
