# Create a bash script that takes a list of AGI loci and references a blast database you create to find their respective cDNA, then export it as a fasta file. 

Imagine you are working in a lab studying iron homeostasis in Arabidopsis thaliana. The AGI loci of 48 genes were identified by a senior lab member who performed a functional analysis on RNA-Seq data from a set of Arabidopsis plants grown in both replete and iron-deficient conditions. For further validation, you are asked to perform qPCR on these genes and will need the cDNA of these genes to design the primers. Write a bash script that takes the list of Arabidopsis Genome Initiative (AGI) loci and references them to the TAIR10 CDNA library, piping the output to a FASTA file for primer design.

The senior lab member has provided you with this list as a text file titled `Potential_Iron_Homeostasis_Genes_Arabidopsis.txt`. It is formatted as below: 

    AT4G19690.1
    AT4G19680.1
    AT1G01580.1
    AT1G01590.1
    AT5G49780.1
    AT1G21620.1
    ...


## Goals: 
1. Write your script in the [`blast.sh`](blast.sh) file located in `module_1_bash_hpc/exercise`.

2. Initialize your variables for the script. There will be 4 - 5 variables:
   * `GENE_LIST` - Location and name of 48 AGI loci as a text file
   * *OPTIONAL* `FASTA_URL` - website link to the FTP server with all of the A. Thaliana cDNA fasta file
   * **FYI:** The `FASTA_URL` should be linked to `FASTA_URL="ftp://ftp.ensemblgenomes.org/pub/plants/release-50/fasta/arabidopsis_thaliana/cdna/Arabidopsis_thaliana.TAIR10.cdna.all.fa.gz"`. Generative AI is likely to provide links that do not function as expected.
   * `FASTA_FILE` - Location and name of the A. Thaliana cDNA fasta file
   * `DB` - name of your database
   * `OUTPUT` - location and name of the output FASTA file. This should be titled `Potential_Iron_Homeostasis_Genes_Arabidopsis_cDNA.fa`.
3. Input the location/name of the file `Potential_Iron_Homeostasis_Genes_Arabidopsis.txt` into a variable called `GENE_LIST`. This file is located in [`Module_1/Resources/Data/Potential_Iron_Homeostasis_Genes_Arabidopsis.txt`](../../../data/examples/Potential_Iron_Homeostasis_Genes_Arabidopsis.txt) and contains the 48 genes' AGI identifier.

4. The cDNA file's location/name should be assigned to the variable `FASTA_FILE`. The URL from the [Ensemble FTP server](https://ftp.ebi.ac.uk/ensemblgenomes/pub/release-61/plants/fasta/arabidopsis_thaliana/) can be assigned to a variable `FASTA_URL`.  
    *Hint: navigate to cdna, and right click the file `Arabidopsis_thaliana.TAIR10.cdna.all.fa.gz`, then copy the ftp link to use for `FASTA_URL'.*   

5.  Fetch the cDNA file `Arabidopsis_thaliana.TAIR10.cdna.all.fa` using the `wget` command. The file will download as a `*.fa.gz` format. You must use `gzip` to extract it to a `.fa` format.  
   *Hint: Retrive the file with* `wget -O $[FASTA_FILE}.gz $FASTA_URL` *within the `module_1_bash_hpc/exercise/` folder. We covered this in M1P4.*
   
6. Use the `makeblastdb` command to create the database `tair10_db`, which was assigned to the variable called `DB`. You can learn about `makeblastdb` and `blast+` [here](https://www.ncbi.nlm.nih.gov/books/NBK279690/).    
   *Hint: Make sure to use parameters `-dbtype nucl`, `-parse_seqids`, and `-out` along with their needed key-value pairs when using the `makeblastdb` command. We covered this in M1P4*

7. Use a while loop to iterate (`read`) each line of `GENE_LIST`, the variable that references `Potential_Iron_Homeostasis_Genes_Arabidopsis.txt`. Store each of these lines as a variable called `AGI`.
    * Within the while loop, use a conditional statement to decide if `AGI` is not empty, as we did in [M1P3](../practice/M1P3_Use_of_Programs_In_Bash.ipynb).
        * Within this conditional block, run the `blastdbcmd` command with parameters `-db "$DB"` and `-entry `AGI`.
        * On the same line as the `makeblastdb` command, append the output file using the `>>` operator to reference the variable `OUTPUT`.
        * Be sure to denote when the while loop is done with `done < $GENE_LIST`.

8. Grant your script executable permissions from the terminal with `chmod +x blast.sh`, and run it with `./blast.sh`.   
   *Hint: Make sure you are in the folder `module_1_bash_hpc/exercise` within the terminal.*
