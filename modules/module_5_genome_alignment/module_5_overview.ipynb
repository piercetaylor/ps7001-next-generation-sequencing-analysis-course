# Module 5 Checklist
## Module 5 - Mapping reads to a Genome and Genome Browsers
This week, we will align reads from 12 *A. thaliana* root samples within [GSE128157](https://www.ncbi.nlm.nih.gov/geo/query/acc.cgi?acc=GSE128157) to the TAIR10 genome using STAR and visualize the results using IGV browser. 

**NOTE: This module includes running programs like STAR and Fastp on large file sets (>10GB), so allot time for Jupyter Notebook sessions in larger increments (4-8 hours) and expect downtime while programs run as you work through the labs, practices, and exercises.**

**Please make sure you have created each Jupyter session with CPU cores = 8 and RAM (memory) >= 32 GB**

## Module Topics:

- File organization
- SRA retrevial 
- Index building for alignment software
- Genome alignment using STAR
- IGV Browsers
- Splicing, Isoforms, and strandedness of reads
