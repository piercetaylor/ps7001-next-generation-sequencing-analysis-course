# Module 7 Checklist
## Module 7 - DAP-Seq and Protein-DNA Interaction Assays
This week, you will learn how DNA Affinity Purification Sequencing (DAP-Seq) and related assays (ChIP-Seq and ATAC-Seq) identify transcription factor binding sites across the genome. You will run a scripted DAP-Seq pipeline on *Arabidopsis* AUXIN RESPONSE FACTOR (ARF) transcription factors (ARF5/MP, ARF2, and ARF16) from the [GSE60141 dataset](https://www.ncbi.nlm.nih.gov/geo/query/acc.cgi?acc=GSE60141), call peaks against a beads-only control, discover motifs, and compare your peaks to the published results in IGV. The paper, [Cistrome and Epicistrome Features Shape the Regulatory DNA Landscape](https://doi.org/10.1016/j.cell.2016.04.038), is available to you on canvas as well.

**NOTE: This module includes running programs like Bowtie2, MACS2, and deepTools on large file sets (>10GB), so allot time for Jupyter Notebook and Hellbender Desktop sessions in larger increments (4-8 hours) and expect downtime while programs run as you work through the labs, practices, and exercises.**

**Please make sure you have created each Session with CPU cores = 8 and RAM (memory) >= 32 GB for use of the practice and exercise**

## Module Topics:

- Transcription factor-DNA interactions, chromatin binding, and motifs
- DAP-Seq, ChIP-Seq, and ATAC-Seq experimental design and controls
- Interpreting DAP-Seq results: peak files, motifs, genome browser tracks, and quality metrics
- Scripted DAP-Seq pipeline: QC, Bowtie2 alignment, and MACS2 peak calling
- Coverage tracks and heatmaps with deepTools
- Motif discovery (HOMER/MEME) and peak annotation
- Visualizing DAP-Seq peaks in IGV
