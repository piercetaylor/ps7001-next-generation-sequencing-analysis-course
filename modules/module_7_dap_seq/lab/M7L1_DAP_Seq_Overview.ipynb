## **Review of Concepts**


### Sequencing by Senthesis (Illumina Sequencing)

*[Third-party figure omitted from the public version; original source not confirmed.]*

Key Steps in Illumina Sequencing:
- **Library Preparation: DNA fragments are prepared with adaptors ligated to both ends**
    - Adaptors contain sequences complementary to oligonucleotides on the flow cell
    - Adaptors also contain primer binding sites for sequencing



- **Cluster Generation:**
    - Fragments bind to the flow cell surface via adaptor sequences
    - Bridge amplification creates clusters of identical DNA molecules
    - Each cluster contains ~1,000 copies of the same fragment



- **Sequencing by Synthesis:**
    - Fluorescently-labeled reversible-terminator nucleotides are added
    - DNA polymerase incorporates one nucleotide at a time
    - After each incorporation, the flow cell is imaged to detect fluorescence
    - The terminator and fluorophore are cleaved, and the cycle repeats
    - Typically generates 75-150 bp reads (single-end or paired-end)



- **Data Output:**
    - Produces FASTQ files containing millions of short sequence reads
    - Each read represents a DNA fragment from the original library

### Transcription Factors and DNA Binding

*[Third-party figure omitted from the public version; original source not confirmed.]*

Transcription factors (TFs) are proteins that regulate gene expression by binding to specific DNA sequences, typically in gene regulatory regions (promoters, enhancers). Transcription factors have several properties that allow them to help modulate the expression (**increasing or inhibiting**) of genes within an organism: 

1. DNA-Binding Domains (DBDs):

- Structured TF(protein) regions that recognize specific DNA sequences
- Common types include: zinc fingers, helix-turn-helix, leucine zippers, basic helix-loop-helix
- Different families of TFs have characteristic DBD structures


2. Sequence Specificity:

- TFs recognize short DNA sequences (typically 6-20 bp) - **A MOTIF**
- Binding is determined by protein-DNA contacts (hydrogen bonds, van der Waals interactions)
- Specificity is not absolute—TFs can bind to similar but not identical sequences

**It is important to realize TFs can be sticky (binding to all DNA) or can require partner proteins to allow for DNA binding** These interactions allow for an organism to dynamically regulate the expression of genes in response to environmental stimuli.


3. Binding Affinity:

- Influenced by the DNA sequence matching to the preferred motif
- Affected by DNA shape and structural features
- Modulated by DNA modifications (e.g., methylation)
- Can involve cooperative binding with other proteins


4. Functional Consequences:
- TF binding can activate or repress transcription
- Recruits or blocks RNA polymerase and co-factors
- Can alter chromatin structure


### Chromatin - Context for the biological relevance of Motifs for TF binding

In eukaryotic cells, DNA does not exist as a naked double helix but is packaged with histone proteins into chromatin.

**Chromatin Structure:**

*[Third-party figure omitted from the public version; original source not confirmed.]*
**DNA-Helix**: The Structure of DNA is a double helix, and this is how it is displayed in models and (generally) in-vitro
**Nucleosomes:** Form a structure of "beads on a string" and help DNA to be further compacted into higher-order structures

**Chromatin States:**

- Euchromatin: Open, accessible, transcriptionally active
- Heterochromatin: Condensed, inaccessible, transcriptionally silent


How does this impact the binding of TFs, and thereby our DAP-Seq protocols:
- TFs can only bind to accessible DNA regions
- Nucleosome positioning can occlude or expose binding sites
- Chromatin remodeling complexes can alter accessibility
- Histone modifications influence chromatin state



### In Vivo vs. In Vitro Binding:
This distinction is crucial for understanding DAP-Seq and ChIP-Seq. When we perform experiments within an organism vs in vitro, the access to potential motifs is changed drastically.

**In vivo (ChIP-Seq):** TF binding occurs in the native chromatin context

- Reflects actual cellular binding landscape
- Limited by chromatin accessibility
- Influenced by protein-protein interactions


**In vitro (DAP-Seq):** TF binding occurs with naked genomic DNA

- Reveals all potential binding sites based on sequence
- Not limited by chromatin accessibility
- Shows intrinsic DNA-binding preferences

In the context of our paper, Bartlett and O'Malley show that integrating DAP-Seq data with chromatin accessibility data (DNase-seq, ATAC-seq) can predict in vivo binding sites across multiple tissues without performing numerous ChIP-seq experiments.


To review, A DNA binding motif is a characteristic sequence pattern that reflects the preferred binding site of a transcription factor (TF), and understanding how these motifs are represented is essential for interpreting TF-DNA interactions in assays such as DAP-seq and ChIP-seq. Motifs can be summarized as a consensus sequence, which captures the most common nucleotide at each position (for example, the bZIP CACGTG motif), but this simplicity often obscures important variability in binding preferences. 

More detailed representations use a position weight matrix (PWM)—also known as a position-specific scoring matrix—which records the frequency of each nucleotide at every base position across many observed binding sites, enabling quantitative scoring of potential binding sequences. **PWMs are often visualized as sequence logos**, graphical displays where the height of each letter indicates both nucleotide frequency and overall information content, clearly illustrating both preferred and tolerated bases:

*[Figure omitted: Bartlett et al. (2017) Nature Protocols 12:1659-1672, [doi:10.1038/nprot.2017.055](https://doi.org/10.1038/nprot.2017.055).]*

Simply, DAP-seq measures TF binding in vitro on naked genomic DNA, capturing the TF’s opportunistic motif binding, whereas ChIP-seq measures TF binding in vivo, where chromatin accessibility strongly influences which motif binds within the cell.

## DAP-Seq Overview: How and Why?

Multicellular organisms encode an enormous number of transcription factors—approximately 2,000 in Arabidopsis and roughly 1,600 in humans—making it a significant challenge to comprehensively map the binding sites required to understand gene regulatory networks. 

Traditional approaches, such as ChIP-seq, were the gold standard for profiling in vivo TF binding. They are powerful but limited: they require high-quality antibodies or transgenic lines expressing epitope-tagged TFs, which are costly, time-consuming. ChIP-seq is low-throughput and often ineffective method for TFs that are lowly expressed or active only in rare cell populations.

Previous to DAP-Seq, high-throughput in vitro methods such as protein-binding microarrays (PBM) and SELEX efficiently identify motifs using synthetic oligonucleotides, but this artificial DNA environment lacks genomic context, including natural sequence complexity and methylation patterns. 

DAP-seq addresses these challenges by combining high-throughput in vitro TF expression with native genomic DNA and next-generation sequencing. Because TFs are purified directly, the method avoids the need for antibodies or transgenic lines, while the use of real genomic DNA preserves natural sequence context and modifications. When paired with NGS, this allows DAP-seq to generate genome-wide, quantitative TF binding profiles in a scalable and efficient manner.

**Key Advantages of DAP-Seq:**
- Scalable to hundreds of TFs
- Uses genomic DNA with native methylation patterns
- No antibodies required
- Relatively inexpensive (~$10-20 per TF)
- Can be applied to any organism with a sequenced genome* (With enough preparation and effort prior... Think Y1H screening + EMSA's to find proteins suitable for DAP-Seq...)

## Steps of DAP-Seq

*[Figure omitted: O'Malley et al. (2016) Cell 165:1280-1292, [doi:10.1016/j.cell.2016.04.038](https://doi.org/10.1016/j.cell.2016.04.038), Fig. 1.]*

**Genomic Library Preparation (A.)**:
The first component of DAP-Seq is preparing a genomic DNA (gDNA) library. This library can be used for many individual DAP-Seq experiments, increasing throughput and reducing batch effects.


*[Figure omitted: Bartlett et al. (2017) Nature Protocols 12:1659-1672, [doi:10.1038/nprot.2017.055](https://doi.org/10.1038/nprot.2017.055).]*

DNA preparation starts with clean genomic DNA fragmented into 200 bp segments. This is ideal for Illumina sequencing and for transcription factor binding. These breaks leave a mix of uneven DNA ends, so the fragments are “fixed” to make all their ends uniform, and a single A base is added to each 3′ end so Illumina adaptors can attach properly. Y-shaped adaptors are then ligated onto the DNA, giving each fragment the sequences needed for sequencing and indexing. After cleaning up the reactions and measuring the final library, QC for adaptor ligation is checked using qPCR or a BioAnalyzer; Properly ligated DNA shows strong amplification and appears slightly longer than unligated DNA. The same library can be PCR-amplified to create methylation-free ampDAP library.

**Transcription Factor Expression and Purification**

In DAP-seq, the transcription factor (TF) protein is produced in vitro with an affinity tag to allow easy purification. Instead of expressing proteins in living cells, DAP-seq uses cell-free transcription/translation systems, which are faster, high-throughput, and work even for TFs that might be toxic to cells. The TF coding sequences are cloned into a Gateway-compatible pIX-HALO vector, which adds an N-terminal HaloTag and includes promoters for different expression systems. In the case of Arabidopsis, over 1,800 TF open reading frames were transferred into this vector to create HaloTag–TF fusion proteins. The HaloTag is a modified enzyme that forms a covalent bond with chloroalkane ligands, allowing extremely stable and essentially irreversible binding. Unlike traditional tags such as FLAG or His, it does not require antibodies and provides a very low background, and ligand-coupled beads are commercially available for purification.

*[Figure omitted: Bartlett et al. (2017) Nature Protocols 12:1659-1672, [doi:10.1038/nprot.2017.055](https://doi.org/10.1038/nprot.2017.055).]*

**DAP - DNA Affinity Purification**

*[Figure omitted: Bartlett et al. (2017) Nature Protocols 12:1659-1672, [doi:10.1038/nprot.2017.055](https://doi.org/10.1038/nprot.2017.055).]*

DNA affinity purification is the central step in DAP-seq where the transcription factor selectively captures its genomic binding sites from the DNA library. The adaptor-ligated genomic DNA, typically 30–100 ng depending on genome size, is added to the bead-bound HaloTag–TF fusion protein and incubated at room temperature to allow sequence-specific binding. High-affinity sites are preferentially captured, while nonspecific DNA remains in solution. Following binding, unbound DNA is removed through multiple stringent washes, which reduce background and increase specificity. Finally, the bound DNA is recovered by heating the beads to denature the TF, releasing the specifically bound fragments into solution. The recovered DNA represents genomic regions recognized by the TF, with the amount depending on the number of binding sites, and is sufficient for downstream PCR amplification and sequencing.

**Sequencing of DAP Library**

Recovered DNA is converted into a sequencing-ready library by PCR, which both amplifies the DNA and adds full-length Illumina adaptors with sample-specific index sequences for multiplexing. Universal and indexed primers bind the adaptor sequences to provide flow-cell attachment and sequencing primer sites. The PCR reaction uses a high-fidelity polymerase to generate enough material for sequencing. This data is then used for downstream analysis.


**Optional: ampDAP-Seq**

We can amplify gDNA library (11 cycles of PCR to create methylation-free library), then Repeat steps 3-5 with the amplified library.

## ChIP-Seq Experiments

ChIP-Seq (Chromatin Immunoprecipitation Sequencing) is the gold standard for mapping in vivo TF binding sites.

*[Third-party figure omitted from the public version; original source not confirmed.]*

**Steps in ChIP-Seq:**
- Crosslinking: Formaldehyde fixes protein-DNA interactions in living cells
- Chromatin fragmentation: Sonication breaks chromatin into ~200-500 bp fragments
- Immunoprecipitation: Antibody against TF pulls down TF-DNA complexes
- Reverse crosslinking: Heat reverses formaldehyde crosslinks
- DNA purification and sequencing: Recovered DNA represents TF-bound regions and is sequenced with Illumina sequencing (or similar)

ChIP-seq measures in vivo interactions; it captures the real cellular context of TF binding, including the effects of chromatin accessibility, histone modifications, and protein–protein interactions where TFs are recruited by partners. It can also provide insights into temporal dynamics, revealing TF binding under specific conditions, and the method is well-established and widely used in molecular biology. However, ChIP-seq also has notable limitations.

Sadly, it is low throughput, as each TF requires a specific antibody or a transgenic line, and the technique presents technical challenges, including efficient crosslinking, sonication, and variability in antibody quality. Many ChIP-seq peaks represent indirect binding, making it difficult to distinguish direct from indirect interactions, and the method is constrained by cell or tissue specificity, requiring the TF to be expressed in the sampled material. As a result, only a limited number of ChIP-seq datasets exist compared to the hundreds of datasets available from high-throughput approaches like DAP-seq.


# ATAC-Seq Experimental Design

ATAC-Seq (Assay for Transposase-Accessible Chromatin with Sequencing) is a genome-wide method to map open chromatin regions. It uses a hyperactive Tn5 transposase that simultaneously cuts accessible DNA and ligates sequencing adaptors in a single "tagmentation" reaction. Open chromatin regions, where nucleosomes are depleted, correspond to active regulatory elements such as promoters and enhancers, whereas closed chromatin is tightly packaged and generally inaccessible to transcription factors. The protocol begins with gentle nuclei isolation to preserve chromatin structure, followed by transposition with Tn5. The tagmented DNA is purified, PCR-amplified with sample indexes, size-selected to enrich nucleosome-free and mono-nucleosome fragments, and sequenced, typically in paired-end mode to infer fragment sizes.  

*[Third-party figure omitted from the public version; original source not confirmed.]*

Data analysis involves peak calling (commonly using `MACS2`) to identify open chromatin regions, examining fragment size distribution to distinguish nucleosome-free (<100 bp) and nucleosome-bound fragments (~200–600 bp), and performing TF footprinting to detect protected sites within peaks where TFs are bound. ATAC-seq offers a simple, fast protocol requiring low input, does not need antibodies, and provides a genome-wide view of regulatory elements. However, it does not directly identify which TFs bind and is less specific than ChIP-seq, requiring computational inference to predict TF occupancy.  

When combined with DAP-seq, ATAC-seq can help predict tissue-specific binding: DAP-seq identifies all potential TF binding sites in a chromatin-free context, while ATAC-seq identifies regions that are actually accessible in specific tissues. Overlaying these datasets highlights likely in vivo binding sites, allowing cost-effective prediction of tissue-specific regulatory activity without performing ChIP-seq for every TF in every tissue.
